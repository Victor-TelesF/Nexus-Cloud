```python
from dashboards.database import engine
import pandas as pd

query = """
    WITH resumo AS(
    SELECT EXTRACT(
        YEAR FROM p.data_pagamento
        ) AS ano,
    EXTRACT(
        MONTH FROM p.data_pagamento
        ) AS mes,
    SUM(p.valor) AS total
    FROM nexus.pagamentos p
    WHERE p.data_vencimento IS NOT NULL AND p.status = 'pago'
    GROUP BY ano, mes
    ORDER BY ano, mes
)
SELECT ano,mes,total,
ROUND(
    ((total / NULLIF(LAG(total, 1, 0) OVER(ORDER BY ano, mes), 0) - 1) * 100), 2
) AS variacao_percentual 
FROM resumo
ORDER BY ano, mes;
"""



clientes = pd.read_sql_table(
    "clientes",
    con=engine,
    schema="nexus"
)

contratos = pd.read_sql_table(
    "contratos",
    con=engine,
    schema="nexus"
)

pagamentos = pd.read_sql_table(
    "pagamentos", 
    con=engine,
    schema="nexus"
)

servicos = pd.read_sql_table(
    "servico",
    con=engine,
    schema="nexus"
)

clientes.info()
pagamentos.info()
contratos.info()
servicos.info()
```
